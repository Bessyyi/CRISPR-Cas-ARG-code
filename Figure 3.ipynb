```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib_venn import venn2, venn2_circles
import numpy as np
from matplotlib import patches
import matplotlib as mpl
```

```python
# 定义集合大小
phage_only = 7257-4333+2068-215+8  # 仅属于phage的数量
plasmid_only = 16325 -4333+868+84  # 仅属于plasmid的数量
both_phage_plasmid = 1397-8  # 同时属于phage和plasmid的数量
unknown = 42741-phage_only-plasmid_only-both_phage_plasmid  # 未知的数量

# 总数据
# total
total = 42741
print(total)
# 创建图形
fig, ax = plt.subplots()

# 创建两个集合的Venn图
v = venn2(subsets=(phage_only, plasmid_only, both_phage_plasmid),
             set_labels=('Phage', 'Plasmid'),set_colors=('#0073C2FF','#EFC000FF'),alpha = 0.8)

for text in v.set_labels:
    text.set_fontsize(10);
    text.set_fontweight("bold")
for text in v.subset_labels:
    text.set_fontsize(10)
    text.set_fontweight("bold")
for subset_id, label in zip(('10', '01', '11'), v.subset_labels):
    if label is not None:
        value = int(label.get_text())
        label.set_text(f"{value:,}")
        label.set_fontsize(10)
        label.set_fontweight("bold")
plt.text(0, 0.6, f"Unknown: {unknown:,}", horizontalalignment='center', fontsize=10, fontweight='bold')
# 添加未知部分的注释
# plt.text(0.5, -0.4, f"Unknown: {unknown}", horizontalalignment='center', fontsize=12)
v=venn2_circles( subsets=(phage_only, plasmid_only, both_phage_plasmid),
        linestyle='--', linewidth=2, color="black"#外框线型、线宽、颜色
       )

# 添加大圈的边框，使Phage和Plasmid包含在内
circle = patches.Circle((0, 0), 0.8, edgecolor='black', linestyle='--',fill=False,zorder=0, linewidth=2.5)
ax.add_patch(circle) 
# 调整图形
# plt.title("Venn Diagram with Unknown Data")
plt.axis('equal')
# plt.savefig("all_spacer_uniq_disctribution_filter_20260504.png",bbox_inches = 'tight',dpi=800)
plt.show()
```
